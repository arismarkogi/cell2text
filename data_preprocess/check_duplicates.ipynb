```python
import pytz
import pandas as pd
from datasets import Dataset, load_from_disk

def check_for_duplicates(dataset1: Dataset, dataset2: Dataset) -> tuple[bool, list[int]]:
    """
    Checks for entries with the same 'soma_joinid' column in two Hugging Face datasets.

    This function is efficient for large datasets as it converts the 'soma_joinid'
    columns to sets and finds the intersection.

    Args:
        dataset1 (Dataset): The first Hugging Face dataset.
        dataset2 (Dataset): The second Hugging Face dataset.

    Returns:
        tuple[bool, list[int]]: A tuple containing:
            - A boolean indicating if any duplicate IDs were found (True if duplicates exist).
            - A list of the duplicate 'soma_joinid' values.
    """
    try:
        # Get the 'soma_joinid' column from each dataset
        ids1 = dataset1["soma_joinid"]
        ids2 = dataset2["soma_joinid"]


        # Convert the lists of IDs to sets for efficient comparison (O(1) lookup time)
        set1 = set(ids1)
        set2 = set(ids2)

        # Find the intersection of the two sets to get the common IDs
        common_ids = set1.intersection(set2)

        # Convert the set of common IDs back to a list for the return value
        duplicate_ids = list(common_ids)
        
        # Check if the list of duplicate IDs is not empty
        has_duplicates = len(duplicate_ids) > 0

        print(f"Number of common 'soma_joinid' entries found: {len(duplicate_ids)}")

        return has_duplicates, duplicate_ids
    except KeyError as e:
        print(f"Error: The 'soma_joinid' column was not found in one of the datasets. {e}")
        return False, []
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return False, []

# --- Example Usage ---
# Please specify the paths to your datasets here.
first_dataset_path = "/home/arism/datasets/train/dataset_1_with_descriptions"
second_dataset_path = "/home/arism/datasets/val/dataset_1_with_descriptions"

print("Loading datasets from disk...")

# Load the datasets from the specified paths.
dataset_a = load_from_disk(first_dataset_path)
dataset_b = load_from_disk(second_dataset_path)


print(dataset_a)

print("\nChecking for duplicates between Dataset A and Dataset B...")
has_dupes, dupes_list = check_for_duplicates(dataset_a, dataset_b)

print("\n--- Results ---")
if has_dupes:
    print(f"Duplicates found: Yes")
    print(f"List of duplicate IDs: {dupes_list}")
else:
    print(f"Duplicates found: No")

```
